**Iris Flower Clustering Project (K-Means & PCA)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import adjusted_rand_score
import warnings
warnings.filterwarnings('ignore')

# Load Iris Dataset
iris = load_iris()
X = iris.data
y_true = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

df = pd.DataFrame(X, columns=feature_names)
df['True_Label'] = y_true

# Scale the data before PCA and K-Means (Best Practice)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data loaded and scaled successfully.")
```

**1. Apply PCA to Reduce Dataset Dimensions**

```python
# Apply PCA to reduce dataset dimensions to 2D for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]

print(f"Explained Variance Ratio by 2 components: {pca.explained_variance_ratio_.sum():.4f}")
```

**2. Apply K-Means (k=3) & Evaluate**

```python
# Apply K-Means (k=3)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X_scaled)
df['Predicted_Cluster'] = y_kmeans

# Evaluate using Adjusted Rand Index
ari = adjusted_rand_score(y_true, y_kmeans)
print(f"Adjusted Rand Index (True Labels vs Predicted): {ari:.4f}")
```

**3. Visualize Clusters & Compare with True Labels**

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot True Labels
sns.scatterplot(
    ax=axes[0], data=df, x='PCA1', y='PCA2', 
    hue='True_Label', palette='viridis', legend='full', alpha=0.8, s=80
)
axes[0].set_title('True Labels (PCA Reduced)')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')

# Modify legend to show class names
handles, labels = axes[0].get_legend_handles_labels()
axes[0].legend(handles=handles, labels=list(target_names), title="Species")

# Plot K-Means Predicted Clusters
sns.scatterplot(
    ax=axes[1], data=df, x='PCA1', y='PCA2', 
    hue='Predicted_Cluster', palette='Set1', legend='full', alpha=0.8, s=80
)

# Add cluster centers
centers_pca = pca.transform(kmeans.cluster_centers_)
axes[1].scatter(centers_pca[:, 0], centers_pca[:, 1], c='black', s=200, marker='X', label='Centroids')

axes[1].set_title('K-Means Predicted Clusters')
axes[1].set_xlabel('Principal Component 1')
axes[1].set_ylabel('Principal Component 2')
axes[1].legend(title="Cluster")

plt.tight_layout()
plt.show()
```

**4. Save the Model**

```python
# Save the models using joblib
joblib.dump(kmeans, 'kmeans_iris_model.pkl')
joblib.dump(pca, 'pca_iris_model.pkl')
joblib.dump(scaler, 'scaler_iris.pkl')
print("Models saved successfully as 'kmeans_iris_model.pkl', 'pca_iris_model.pkl', and 'scaler_iris.pkl'.")
```
